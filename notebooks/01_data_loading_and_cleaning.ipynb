# 1. Data Loading and Cleaning

## 1.1. Introduction

This notebook focuses on the initial and most crucial step of any data science project: data collection and preparation. We will fetch historical price data for Bitcoin (BTC-USD) and the S&P 500 (^GSPC) using the `yfinance` library. Once downloaded, we will perform essential cleaning steps to ensure the dataset is robust and ready for analysis. This includes handling missing values and structuring the data appropriately. The cleaned data will be saved to a CSV file in the `data/processed` directory for easy access in subsequent notebooks.

## 1.2. Import Libraries

```python
import yfinance as yf
import pandas as pd
import os
```

## 1.3. Define Parameters

Here, we define the tickers for the assets we want to analyze (Bitcoin and S&P 500), the time period for our analysis, and the file path where the processed data will be saved.

```python
ASSETS = ['BTC-USD', '^GSPC']
START_DATE = '2015-01-01'
END_DATE = '2024-01-01'

PROCESSED_DATA_DIR = '../data/processed'
PROCESSED_DATA_FILE = os.path.join(PROCESSED_DATA_DIR, 'btc_sp500_data.csv')

# Ensure the processed data directory exists
if not os.path.exists(PROCESSED_DATA_DIR):
    os.makedirs(PROCESSED_DATA_DIR)
```

## 1.4. Download Data

We use the `yfinance.download()` function to fetch the historical 'Adj Close' prices for our selected assets over the specified date range. The 'Adj Close' price is used as it accounts for dividends and stock splits, providing a more accurate representation of returns.

```python
data = yf.download(ASSETS, start=START_DATE, end=END_DATE)['Adj Close']
```

## 1.5. Inspect and Clean Data

Real-world data is rarely perfect. We'll inspect the downloaded data for missing values, which can occur on weekends, holidays, or due to data feed issues. We will use forward-fill (`ffill`) to handle these `NaN` values, carrying over the last known price to fill the gaps. This method is suitable for time-series data as it assumes the price remains constant when no new information is available.

```python
print("Initial Data Head:")
print(data.head())
print("
Missing Values Before Cleaning:")
print(data.isnull().sum())
```

```python
# Forward-fill missing values
data.fillna(method='ffill', inplace=True)

print("
Missing Values After Cleaning:")
print(data.isnull().sum())
```

## 1.6. Save Processed Data

Finally, we save the cleaned and processed data to a CSV file. This allows us to easily load the data in other notebooks without needing to re-download and re-clean it each time, ensuring consistency across our analyses.

```python
data.to_csv(PROCESSED_DATA_FILE)
print(f"
Data saved to {PROCESSED_DATA_FILE}")
```
